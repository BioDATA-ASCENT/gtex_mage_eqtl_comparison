Load fine mapping results

```r
fine <- fread("/grid/chambwe/home/founta/code_athon/data/expression/eQTL_summary.MAGE.v1.0.txt")
head(fine)
```

```r
fine2 <- fine[which(fine$finemappedCredibleSets > 0),]
dim(fine2)
head(fine2)
```

Keep only genes with finemapped resutls from gene expression matrix

```r
dms2 <- unique(dms[which(dms$ID %in% fine2$ensemblID),])
dim(dms2)
head(dms2)
```

```r
table(dms2$`#Chr`)
```

```r
length(unique(dms2$ID))
```

```r
dms3 <- dms2[which(dms2$`#Chr` == "chr1"),]
dms3 <- dms3[,-c(1:3)]

dim(dms3)
head(dms3)
```

```r
colnames(dms3)[1] <- "gene"
head(dms3)
```

Move site to row_names and transpose matrix

```r
rownames(dms3) <- NULL
sig_mat <- dms3 %>% column_to_rownames("gene")
 
sig_t <- as.data.frame(t(sig_mat))
dim(sig_t)
head(sig_t)
```

Create individual and family ID

```r
sig_t <- sig_t %>%  
  rownames_to_column("IID") %>%  
  mutate(FID = IID, .before = IID)

sig_t$FID <- "0"

dim(sig_t)   # expect 50 samples x (2 + 8757 sites)  
head(sig_t[,1:5])  
```

Missing values

```r
sum(is.na(sig_t))
```

Save file with and without header

```r
write.table(  
  sig_t,  
  file = "~/code_athon/metadata/1kgp_phenotype_chr1.txt",  
  sep = "\t",  
  row.names = FALSE,  
  col.names = FALSE,  
  quote = FALSE,  
  na = "NA"  
)

write.table(  
  sig_t,  
  file = "~/code_athon/metadata/1kgp_phenotype_phenotype_with_header_chr1.txt",  
  sep = "\t",  
  row.names = FALSE,  
  col.names = TRUE,  
  quote = FALSE,  
  na = "NA"  
)  

write.table(  
  sig_t$IID,  
  file = "~/code_athon/metadata/expr_samples.txt",  
  sep = "\t",  
  row.names = FALSE,  
  col.names = FALSE,  
  quote = FALSE,  
  na = "NA"  
)  
```

Save site-to-mpheno index

```r
site_order <- colnames(sig_t)[-(1:2)]  
write.csv(  
  data.frame(mpheno = seq_along(site_order), site = site_order),  
  "~/code_athon/metadata/1kgp_gene_mpheno_index_chr1.csv",  
  row.names = FALSE  
)  
```

## Create covar file

Read file

```r
library(data.table)
pheno <- fread("~/code_athon/data/demographic/sample.metadata.MAGE.v1.0.txt")
head(pheno)
```

```r
table(pheno$continentalGroup)
```

```r
# GCTA convention for discrete covariates like sex: recode as 1/2 (not letters)  
# (PLINK/GCTA convention: 1 = male, 2 = female — but any consistent numeric coding works)  
covar <- pheno %>%  
  mutate(  
    FID = "0",  
    IID = sample_kgpID,  
    SEX = case_when(  
      sex == "XY" ~ "M",  
      sex == "XX" ~ "F",  
      TRUE ~ NA_character_   # anything unexpected becomes NA  
    )  
  ) %>%  
  select(FID, IID, SEX)

head(covar)
```

```r
table(covar$SEX)
```

Save file

```r

write.table(  
  covar,  
  file = "/grid/chambwe/home/founta/code_athon/metadata/sex_covar.txt",  
  sep = "\t",  
  row.names = FALSE,  
  col.names = FALSE,  
  quote = FALSE  
)  
```
